````python
# ─── Ejercicio 4 — Filtrado de filas ────────────────────────────────────────
# Filtrar es quedarse solo con las filas que cumplen una condición. Cuando
# hay varias condiciones, cada una debe expresarse por separado y después
# combinarlas (todas a la vez, o al menos una). Ten en cuenta que la columna
# 'activo' contiene valores verdadero/falso, no texto. Trabaja con
# 'df_usuarios'.
#
# 1. Guarda en 'usuarios_populares' los usuarios con MÁS de 1000 seguidores.
# 2. Guarda en 'populares_las_palmas' los usuarios de la ciudad
#    'Las Palmas' que ADEMÁS tengan más de 1000 seguidores (deben cumplirse
#    las dos condiciones a la vez).
# 3. Guarda en 'jovenes_o_influyentes' los usuarios que cumplan AL MENOS
#    UNA de estas condiciones: tener menos de 25 años o tener más de 10000
#    seguidores.
# 4. Guarda en 'creadores_activos' los usuarios con tipo de cuenta
#    'creator' cuya cuenta esté activa.
# 5. Guarda en 'populares_las_palmas_texto' el mismo resultado del paso 2,
#    pero escribiendo la condición completa como una única expresión de
#    texto con el método de Pandas que lo permite. Comprueba mostrando por
#    pantalla que ambos filtros devuelven el mismo número de filas.
#    Por ejemplo, con esta tabla de 3 usuarios (esto NO es la solución):
#           nombre  ciudad  seguidores
#      0       Ana  Madrid        5000
#      1     Borja  Bilbao         900
#      2    Carmen  Madrid         800
#    la condición "ciudad Madrid Y más de 1000 seguidores" devolvería solo
#    la fila de Ana, y la condición "ciudad Bilbao O más de 4000
#    seguidores" devolvería las filas de Ana y de Borja. Fíjate en que las
#    filas resultantes conservan su etiqueta de índice original.

usuarios_populares = df_usuarios[df_usuarios['seguidores'] > 1000]
populares_las_palmas = df_usuarios[(df_usuarios['ciudad'] == 'Las Palmas') & (df_usuarios['seguidores'] > 1000)]
jovenes_o_influyentes = df_usuarios[(df_usuarios['edad'] < 25) | (df_usuarios['seguidores'] > 10000)]
creadores_activos = df_usuarios[(df_usuarios['tipo_cuenta'] == 'creator') & (df_usuarios['activo'])]
populares_las_palmas_texto = df_usuarios.query("ciudad == 'Las Palmas' and seguidores > 1000")
print(len(usuarios_populares), len(populares_las_palmas), len(jovenes_o_influyentes), len(creadores_activos))
print('Mismo número de filas:', len(populares_las_palmas) == len(populares_las_palmas_texto))
````

````python
# ─── Ejercicio 5 — Valores nulos ────────────────────────────────────────────
# Algunos usuarios y publicaciones tienen campos vacíos (valores nulos). Hay
# que localizarlos y decidir qué hacer con ellos antes de analizar los datos.
# IMPORTANTE: las tablas originales ('df_usuarios' y 'df_publicaciones')
# no deben modificarse: trabaja siempre sobre copias.
#
# 1. Guarda en 'nulos_por_columna' el número de valores nulos de cada
#    columna de 'df_usuarios'.
#    Por ejemplo, para una tabla con las columnas 'edad' (2 nulos) y
#    'ciudad' (0 nulos), el RESULTADO indicaría: edad 2, ciudad 0.
# 2. Guarda en 'usuarios_con_edad' las filas de 'df_usuarios' cuya 'edad'
#    NO es nula (es decir, elimina solo las filas con nulo en esa columna;
#    los nulos de otras columnas no importan).
# 3. Guarda en 'usuarios_limpio' una COPIA de 'df_usuarios' en la que los
#    nulos se hayan sustituido así:
#      - 'edad'   -> la mediana de la propia columna 'edad'
#      - 'ciudad' -> el texto 'Desconocido'
#      - 'bio'    -> el texto 'Desconocido'
#    Por ejemplo, para las edades 20, nulo, 40, 60 la mediana es 40 y el
#    RESULTADO sería 20, 40, 40, 60 (esto NO es la solución).
# 4. Guarda en 'publicaciones_limpio' una COPIA de 'df_publicaciones' en la
#    que los nulos de 'likes' y de 'comentarios' se hayan sustituido por la
#    mediana de cada una de esas dos columnas.
# 5. Muestra por pantalla el total de nulos de 'usuarios_limpio' y de
#    'publicaciones_limpio' (deben ser 0) y el de 'df_usuarios' (debe seguir
#    siendo mayor que 0: así compruebas que el original no se ha
#    modificado).

nulos_por_columna = df_usuarios.isnull().sum()
usuarios_con_edad = df_usuarios.dropna(subset=['edad'])

usuarios_limpio = df_usuarios.copy()
usuarios_limpio['edad'] = usuarios_limpio['edad'].fillna(usuarios_limpio['edad'].median())
usuarios_limpio['ciudad'] = usuarios_limpio['ciudad'].fillna('Desconocido')
usuarios_limpio['bio'] = usuarios_limpio['bio'].fillna('Desconocido')

publicaciones_limpio = df_publicaciones.copy()
publicaciones_limpio['likes'] = publicaciones_limpio['likes'].fillna(publicaciones_limpio['likes'].median())
publicaciones_limpio['comentarios'] = publicaciones_limpio['comentarios'].fillna(publicaciones_limpio['comentarios'].median())

print('Nulos usuarios_limpio:', usuarios_limpio.isnull().sum().sum())
print('Nulos publicaciones_limpio:', publicaciones_limpio.isnull().sum().sum())
print('Nulos df_usuarios (original):', df_usuarios.isnull().sum().sum())
````

````python
# ─── Ejercicio A.1 — Auditoría de una tabla ─────────────────────────────────
# Una auditoría resume la calidad de una tabla con una fila por cada una de
# sus columnas.
#
# 1. Define una función 'auditar_dataframe(df)' que reciba una tabla y
#    devuelva una NUEVA tabla con una fila por cada columna de 'df' y estas
#    cinco columnas:
#      'columna'          -> nombre de la columna auditada
#      'tipo_dato'        -> tipo de dato de la columna, escrito como texto
#      'n_nulos'          -> número de valores nulos
#      'porcentaje_nulos' -> porcentaje de nulos sobre el total de filas,
#                            redondeado a 2 decimales
#      'n_distintos'      -> número de valores distintos (sin contar nulos)
#    Por ejemplo, para esta tabla de 4 filas (esto NO es la solución):
#           edad  ciudad
#      0    20.0  Madrid
#      1     NaN  Bilbao
#      2    30.0  Madrid
#      3    30.0  Madrid
#    la auditoría devolvería:
#        columna  tipo_dato  n_nulos  porcentaje_nulos  n_distintos
#      0    edad    float64        1              25.0            2
#      1  ciudad        str        0               0.0            2
# 2. Guarda en 'auditoria_usuarios' y en 'auditoria_publicaciones' el
#    resultado de auditar 'df_usuarios' y 'df_publicaciones'.
# 3. Guarda en 'columnas_con_nulos' una lista de Python con los nombres de
#    las columnas de 'df_usuarios' que tienen al menos un nulo, obtenida a
#    partir de la auditoría (no volviendo a contar los nulos a mano).
# 4. Guarda en 'columna_mas_nulos_publicaciones' el nombre de la columna de
#    'df_publicaciones' con mayor porcentaje de nulos, también a partir de
#    la auditoría.

def auditar_dataframe(df):
    filas = []
    for columna in df.columns:
        n_nulos = int(df[columna].isnull().sum())
        filas.append({
            'columna': columna,
            'tipo_dato': str(df[columna].dtype),
            'n_nulos': n_nulos,
            'porcentaje_nulos': round(n_nulos / len(df) * 100, 2),
            'n_distintos': int(df[columna].nunique()),
        })
    return pd.DataFrame(filas)


auditoria_usuarios = auditar_dataframe(df_usuarios)
auditoria_publicaciones = auditar_dataframe(df_publicaciones)
columnas_con_nulos = auditoria_usuarios[auditoria_usuarios['n_nulos'] > 0]['columna'].tolist()
columna_mas_nulos_publicaciones = auditoria_publicaciones.sort_values('porcentaje_nulos', ascending=False).iloc[0]['columna']
print(auditoria_usuarios)
print(columnas_con_nulos, columna_mas_nulos_publicaciones)
````

````python
# ─── Ejercicio A.2 — Limpieza robusta de usuarios ───────────────────────────
# Los datos de texto llegan a veces con espacios sobrantes o con mayúsculas
# y minúsculas inconsistentes, y las tablas reales tienen nulos. Vas a
# preparar una función que deje limpia la tabla de usuarios sin modificar la
# original.
#
# 1. Define una función 'limpiar_usuarios_avanzado(df)' que reciba la tabla
#    de usuarios y devuelva una COPIA limpia con estas reglas:
#      - En las columnas 'nombre', 'apellidos' y 'ciudad': elimina los
#        espacios sobrantes del principio y del final y deja cada palabra
#        con la inicial en mayúscula y el resto en minúscula.
#      - Los nulos de las columnas de texto se sustituyen por 'Desconocido'.
#      - Los nulos de las columnas numéricas se sustituyen por la mediana de
#        su propia columna.
#      - La columna 'fecha_registro' (si existe) se convierte a tipo fecha.
#      - Las columnas verdadero/falso no se tocan.
#    La función debe seguir funcionando si a la tabla le falta alguna de
#    esas columnas, y debe ser IDEMPOTENTE: limpiar una tabla que ya está
#    limpia no debe cambiar nada.
#    Por ejemplo, para esta tabla (esto NO es la solución):
#             nombre       ciudad  edad
#      0      ' ana '  ' Madrid '  25.0
#      1         None     'bilbao'   NaN
#      2     'CARMEN'         None  41.0
#    el resultado sería:
#             nombre       ciudad  edad
#      0         Ana       Madrid  25.0
#      1 Desconocido       Bilbao  33.0
#      2      Carmen  Desconocido  41.0
# 2. Guarda en 'usuarios_limpios' el resultado de limpiar 'df_usuarios'.
# 3. Guarda en 'mismas_dimensiones' un valor True/False que indique si la
#    tabla limpia tiene las mismas filas y columnas que la original.
# 4. Guarda en 'nulos_tras_limpiar' el número total de nulos de la tabla
#    limpia (debería ser 0). Comprueba también, mostrándolo por pantalla,
#    que 'df_usuarios' NO se ha modificado y sigue teniendo nulos.
# 5. Comprueba que la limpieza funciona con datos sucios: guarda en
#    'usuarios_sucios' una copia de 'df_usuarios' en la que hayas ensuciado
#    a mano los 3 primeros nombres (con espacios alrededor y en mayúsculas
#    o minúsculas mezcladas). Guarda en 'nombres_limpios' una lista con los
#    3 primeros nombres de la tabla 'usuarios_sucios' ya limpiada, y en
#    'nombres_coinciden' si esa lista es igual a los 3 primeros nombres de
#    'df_usuarios'.
# 6. Guarda en 'es_idempotente' un valor True/False que indique si limpiar
#    'usuarios_limpios' otra vez da una tabla idéntica a 'usuarios_limpios'.

def limpiar_usuarios_avanzado(df):
    limpio = df.copy()
    columnas_titulo = ['nombre', 'apellidos', 'ciudad']
    for columna in limpio.columns:
        serie = limpio[columna]
        if pd.api.types.is_bool_dtype(serie):
            continue
        if pd.api.types.is_numeric_dtype(serie):
            limpio[columna] = serie.fillna(serie.median())
        elif columna == 'fecha_registro':
            limpio[columna] = pd.to_datetime(serie)
        else:
            serie = serie.str.strip()
            if columna in columnas_titulo:
                serie = serie.str.title()
            limpio[columna] = serie.fillna('Desconocido')
    return limpio


usuarios_limpios = limpiar_usuarios_avanzado(df_usuarios)
mismas_dimensiones = usuarios_limpios.shape == df_usuarios.shape
nulos_tras_limpiar = int(usuarios_limpios.isnull().sum().sum())
print('Nulos original:', df_usuarios.isnull().sum().sum(), '| nulos limpia:', nulos_tras_limpiar)

usuarios_sucios = df_usuarios.copy()
usuarios_sucios.loc[0, 'nombre'] = '  ANA '
usuarios_sucios.loc[1, 'nombre'] = ' pABLO'
usuarios_sucios.loc[2, 'nombre'] = 'miguel  '
nombres_limpios = limpiar_usuarios_avanzado(usuarios_sucios)['nombre'].head(3).tolist()
nombres_coinciden = nombres_limpios == df_usuarios['nombre'].head(3).tolist()
print(nombres_limpios, nombres_coinciden)

es_idempotente = limpiar_usuarios_avanzado(usuarios_limpios).equals(usuarios_limpios)
print('Idempotente:', es_idempotente)
````

````python
# ─── Ejercicio A.3 — Comparación de estrategias para los nulos ──────────────
# No hay una única forma de tratar los nulos de una columna numérica. Vas a
# comparar tres estrategias y decidir cuál es más razonable.
#
# 1. Define una función 'comparar_imputaciones(df, columna)' que reciba una
#    tabla y el nombre de una columna numérica, y devuelva una NUEVA tabla
#    con una fila por estrategia y estas columnas:
#      'estrategia'         -> 'eliminar filas', 'rellenar con media' o
#                              'rellenar con mediana' (en ese orden)
#      'n_filas'            -> filas que conserva la tabla con esa estrategia
#      'media_resultante'   -> media de la columna tras aplicar la estrategia
#      'mediana_resultante' -> mediana de la columna tras aplicar la estrategia
#    La función NO debe modificar la tabla original.
#    Por ejemplo, para una columna con los valores 10, nulo, 20 y 90 (esto
#    NO es la solución):
#                 estrategia  n_filas  media_resultante  mediana_resultante
#      0      eliminar filas        3              40.0                20.0
#      1  rellenar con media        4              40.0                30.0
#      2 rellenar con mediana        4              35.0                20.0
# 2. Guarda en 'comparacion_edad' el resultado de comparar las estrategias
#    en la columna 'edad' de 'df_usuarios'.
# 3. Guarda en 'comparacion_likes' el resultado de comparar las estrategias
#    en la columna 'likes' de 'df_publicaciones'.
# 4. Escribe en un comentario, debajo de las variables, cuál de las tres
#    estrategias te parece más adecuada para cada columna y por qué (fíjate
#    en cuántas filas se pierden y en cómo cambian la media y la mediana).

def comparar_imputaciones(df, columna):
    serie = df[columna]
    sin_nulos = serie.dropna()
    con_media = serie.fillna(serie.mean())
    con_mediana = serie.fillna(serie.median())
    return pd.DataFrame({
        'estrategia': ['eliminar filas', 'rellenar con media', 'rellenar con mediana'],
        'n_filas': [len(sin_nulos), len(con_media), len(con_mediana)],
        'media_resultante': [sin_nulos.mean(), con_media.mean(), con_mediana.mean()],
        'mediana_resultante': [sin_nulos.median(), con_media.median(), con_mediana.median()],
    })


comparacion_edad = comparar_imputaciones(df_usuarios, 'edad')
comparacion_likes = comparar_imputaciones(df_publicaciones, 'likes')
print(comparacion_edad)
print(comparacion_likes)

# Tu razonamiento:
# Eliminar filas pierde datos (5 usuarios y 8 publicaciones) sin necesidad, porque
# los nulos son pocos. Rellenar con la media conserva exactamente la media original,
# pero es sensible a valores extremos; en 'likes', con valores muy dispares, la
# mediana es más robusta. Para 'edad' cualquiera de las dos es razonable; me quedo
# con la mediana en ambas columnas porque no se ve afectada por extremos.
````

````python
# ─── Ejercicio A.4 — Índice con etiquetas repetidas ─────────────────────────
# Un índice no tiene por qué ser único: muchos usuarios comparten nombre.
# Vas a comprobar cómo se comporta la selección por ETIQUETA y por POSICIÓN
# cuando hay etiquetas repetidas. Trabaja con 'df_usuarios'.
#
# 1. Guarda en 'usuarios_por_nombre' una copia de 'df_usuarios' en la que
#    la columna 'nombre' sea el índice.
# 2. Guarda en 'filas_pablo' todas las filas cuya etiqueta de índice es
#    'Pablo', y en 'n_filas_pablo' cuántas son.
#    Por ejemplo, con esta tabla indexada por nombre (esto NO es la
#    solución):
#              edad
#      nombre
#      Ana       20
#      Borja     30
#      Ana       40
#    seleccionar la etiqueta 'Ana' devolvería 2 filas (las de edad 20 y 40)
#    y seleccionar 'Borja' devolvería 1.
# 3. Guarda en 'filas_posicion_10_14' las filas que ocupan las posiciones
#    10 a 14 (ambas incluidas), seleccionadas por POSICIÓN.
# 4. Guarda en 'n_nombres_distintos' el número de etiquetas distintas del
#    índice, y en 'hay_nombres_repetidos' un valor True/False que indique si
#    existen etiquetas repetidas (compara con el número de filas).
# 5. Guarda en 'usuarios_ordenados_por_nombre' la tabla ordenada
#    alfabéticamente según su índice.
# 6. Escribe en un comentario por qué seleccionar la etiqueta 'Pablo'
#    devuelve varias filas mientras que seleccionar una posición devuelve
#    siempre una sola.

usuarios_por_nombre = df_usuarios.set_index('nombre')
filas_pablo = usuarios_por_nombre.loc[['Pablo']]
n_filas_pablo = len(filas_pablo)
filas_posicion_10_14 = usuarios_por_nombre.iloc[10:15]
n_nombres_distintos = len(usuarios_por_nombre.index.unique())
hay_nombres_repetidos = n_nombres_distintos < len(usuarios_por_nombre)
usuarios_ordenados_por_nombre = usuarios_por_nombre.sort_index()
print(n_filas_pablo, filas_posicion_10_14.shape, n_nombres_distintos, hay_nombres_repetidos)

# Tu razonamiento:
# Una etiqueta de índice puede repetirse, así que seleccionar 'Pablo' devuelve todas
# las filas que la llevan (4 aquí). La posición, en cambio, identifica siempre a una
# única fila, porque cada fila ocupa exactamente un lugar (0, 1, 2...).
````

````python
# ─── Ejercicio A.5 — Control de calidad ─────────────────────────────────────
# Un control de calidad comprueba unas reglas de negocio sobre cada fila y
# cuenta los errores por regla.
#
# 1. Define una función 'validar_calidad_usuarios(df)' que reciba la tabla
#    de usuarios y devuelva una tabla de valores verdadero/falso con las
#    mismas filas (y el mismo índice) que 'df' y una columna por regla. Un
#    valor verdadero significa que esa fila INCUMPLE la regla:
#      'edad_nula'                    -> la edad es nula
#      'edad_fuera_de_rango'          -> la edad es menor de 18 o mayor de 55
#      'mas_seguidos_que_seguidores'  -> sigue a más cuentas de las que le
#                                        siguen a él
#      'fila_duplicada'               -> la fila repite por completo otra
#                                        que ya había aparecido antes
#      'id_repetido'                  -> su 'id_usuario' ya había aparecido
#                                        en una fila anterior
#    Por ejemplo, para esta tabla (esto NO es la solución):
#         id_usuario  edad  seguidores  seguidos
#      0           1  25.0         100        10
#      1           2   NaN          50        80
#      2           2  90.0          10         5
#    el resultado sería (verdadero = incumple):
#         edad_nula  edad_fuera_de_rango  mas_seguidos_que_seguidores  fila_duplicada  id_repetido
#      0      False                False                        False           False        False
#      1       True                False                         True           False        False
#      2      False                 True                        False           False         True
# 2. Guarda en 'matriz_calidad' el resultado de validar 'df_usuarios'.
# 3. Guarda en 'errores_por_regla' el número de filas que incumplen cada
#    regla (uno por columna de la matriz anterior).
# 4. Guarda en 'registros_problematicos' las filas de 'df_usuarios' que
#    incumplen AL MENOS UNA regla.
# 5. Guarda en 'publicaciones_sospechosas' las filas de 'df_publicaciones'
#    con MÁS comentarios que 'likes' (los nulos no cuentan como error).
# 6. Comprueba que tu función detecta duplicados: guarda en
#    'errores_con_duplicados' el número de errores por regla que resulta de
#    validar la tabla formada por 'df_usuarios' seguida de sus 2 primeras
#    filas repetidas (deberían aparecer 2 filas duplicadas y 2 ids
#    repetidos).

def validar_calidad_usuarios(df):
    resultado = pd.DataFrame(index=df.index)
    resultado['edad_nula'] = df['edad'].isnull()
    resultado['edad_fuera_de_rango'] = (df['edad'] < 18) | (df['edad'] > 55)
    resultado['mas_seguidos_que_seguidores'] = df['seguidos'] > df['seguidores']
    resultado['fila_duplicada'] = df.duplicated()
    resultado['id_repetido'] = df.duplicated(subset=['id_usuario'])
    return resultado


matriz_calidad = validar_calidad_usuarios(df_usuarios)
errores_por_regla = matriz_calidad.sum()
registros_problematicos = df_usuarios[matriz_calidad.sum(axis=1) > 0]
publicaciones_sospechosas = df_publicaciones[df_publicaciones['comentarios'] > df_publicaciones['likes']]
errores_con_duplicados = validar_calidad_usuarios(pd.concat([df_usuarios, df_usuarios.head(2)])).sum()
print(errores_por_regla)
print(len(registros_problematicos), len(publicaciones_sospechosas))
print(errores_con_duplicados)
````

````python
# ─── Ejercicio A.6 — Integración de tablas: red de contactos e interacciones ───
# La tabla 'df_conexiones' guarda, para cada usuario, la lista de ids de los
# usuarios a los que sigue. Vas a incorporar esa información a las otras
# tablas y a resumirla.
#
# 1. Guarda en 'conexiones_extendidas' una COPIA de 'df_conexiones' con una
#    nueva columna 'n_siguiendo' que contenga, para cada usuario, cuántos
#    ids hay en su lista de la columna 'siguiendo' (tendrás que recorrer
#    las filas o aplicar una operación a cada una).
#    Por ejemplo, para la lista [4, 26, 42] el RESULTADO de la nueva columna
#    sería 3 (esto NO es la solución).
# 2. Guarda en 'ids_alineados' un valor True/False que indique si
#    'df_usuarios' y 'df_conexiones' tienen el mismo 'id_usuario' en cada
#    fila y en el mismo orden (necesario para poder combinarlas fila a
#    fila).
# 3. Guarda en 'usuarios_conexiones' una COPIA de 'df_usuarios' con la
#    columna 'n_siguiendo' de 'conexiones_extendidas' añadida como nueva
#    columna.
# 4. Guarda en 'top_5_conectados' los 5 usuarios que siguen a más cuentas,
#    mostrando solo las columnas 'nombre', 'ciudad' y 'n_siguiendo',
#    ordenados de mayor a menor (si hay empates, conserva el orden de
#    aparición en la tabla).
# 5. Guarda en 'media_siguiendo' la media de 'n_siguiendo' y en
#    'n_sobre_la_media' cuántos usuarios siguen a MÁS cuentas que esa media.
# 6. Guarda en 'top_3_interacciones' las 3 publicaciones con más
#    interacciones, mostrando solo las columnas 'id_publicacion',
#    'tipo_contenido' e 'interacciones'. Las interacciones de una
#    publicación son la suma de sus 'likes', 'comentarios' y 'compartidos'
#    (considera 0 los 'likes' y 'comentarios' nulos). Trabaja sobre una
#    COPIA de 'df_publicaciones'.
#    Por ejemplo, una publicación con 100 likes, 20 comentarios y 5
#    compartidos tendría 125 interacciones.

conexiones_extendidas = df_conexiones.copy()
conexiones_extendidas['n_siguiendo'] = [len(lista) for lista in conexiones_extendidas['siguiendo']]

ids_alineados = bool((df_usuarios['id_usuario'] == df_conexiones['id_usuario']).all())

usuarios_conexiones = df_usuarios.copy()
usuarios_conexiones['n_siguiendo'] = conexiones_extendidas['n_siguiendo']

top_5_conectados = usuarios_conexiones.nlargest(5, 'n_siguiendo')[['nombre', 'ciudad', 'n_siguiendo']]

media_siguiendo = usuarios_conexiones['n_siguiendo'].mean()
n_sobre_la_media = len(usuarios_conexiones[usuarios_conexiones['n_siguiendo'] > media_siguiendo])

publicaciones_interacciones = df_publicaciones.copy()
publicaciones_interacciones['interacciones'] = (
    publicaciones_interacciones['likes'].fillna(0)
    + publicaciones_interacciones['comentarios'].fillna(0)
    + publicaciones_interacciones['compartidos']
)
top_3_interacciones = publicaciones_interacciones.nlargest(3, 'interacciones')[['id_publicacion', 'tipo_contenido', 'interacciones']]
print(ids_alineados, round(media_siguiendo, 2), n_sobre_la_media)
print(top_5_conectados)
print(top_3_interacciones)
````
