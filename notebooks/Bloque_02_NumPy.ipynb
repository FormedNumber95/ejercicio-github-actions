````python
# ─── Ejercicio 5 — Estadísticas por ejes y argmax ──────────────────────────
# 1. Crea manualmente una matriz 2D de NumPy llamada 'publicaciones_ejercicio'
#    con 4 usuarios (filas) x 3 tipos de contenido (columnas: foto, video,
#    texto). Inventa tú el número de publicaciones de cada tipo que ha hecho
#    cada usuario (enteros >= 0). Por ejemplo, con otro tamaño y otros
#    datos, así sería el RESULTADO final (esto NO es la solución, inventa
#    tus propios valores con arrays de Numpy de 4x3):
#      [[3, 0, 1],
#       [1, 2, 0]]

publicaciones_ejercicio = np.array([
    [5, 2, 8],
    [1, 9, 0],
    [4, 4, 4],
    [7, 1, 3],
])

# 2. Calcula el total de publicaciones de CADA USUARIO y guárdalo en 'total_por_usuario'.
# 3. Calcula el total de publicaciones de CADA TIPO DE CONTENIDO y guárdalo en 'total_por_tipo'.
# 4. Obten el ÍNDICE (fila) sobre 'total_por_usuario' del usuario con más publicaciones en total.
#    Guárdalo en'usuario_mas_publicaciones'.

total_por_usuario = publicaciones_ejercicio.sum(axis=1)
total_por_tipo = publicaciones_ejercicio.sum(axis=0)
usuario_mas_publicaciones = int(np.argmax(total_por_usuario))
print(total_por_usuario, total_por_tipo, usuario_mas_publicaciones)
````

````python
# ─── Ejercicio 6 — Apilar y dividir arrays ─────────────────────────────────
# 1. Crea manualmente dos matrices 2D de NumPy llamadas 'grupo_a' y
#    'grupo_b', cada una de 2 filas y 2 columnas (columnas: seguidores,
#    publicaciones), que representen a dos grupos de usuarios nuevos.
#    Inventa tú los valores (enteros >= 0).
#    Por ejemplo, con otros datos, así sería el RESULTADO final (esto NO es
#    la solución, inventa tus propios valores con arrays de Numpy de 2x2):
#      grupo_a -> [[100, 5], [250, 12]]
#      grupo_b -> [[80, 3], [500, 40]]

grupo_a = np.array([[100, 5], [250, 12]])
grupo_b = np.array([[80, 3], [500, 40]])

# 2. Une 'grupo_a' y 'grupo_b' verticalmente (una fila debajo de otra) con
#    metodos especificos de numpy. Guarda el resultado (4 filas
#    x 2 columnas) en 'usuarios_unidos'.
# 3. Divide 'usuarios_unidos' otra vez en dos partes de 2 filas cada una con
#    metodos especificos de numpy, y guárdalas en 'parte_1' y 'parte_2'.
# 4. Comprueba con metodo de numpy de comparación si 'parte_1' coincide con 'grupo_a' y
#    guarda el resultado (True/False) en 'division_correcta'.

usuarios_unidos = np.vstack([grupo_a, grupo_b])
parte_1, parte_2 = np.vsplit(usuarios_unidos, 2)
division_correcta = np.array_equal(parte_1, grupo_a)
print(usuarios_unidos, parte_1, parte_2, division_correcta)
````

````python
# ─── Ejercicio 7 — Funciones universales (frompyfunc) ──────────────────────
# 1. Crea manualmente un array de NumPy llamado 'publicaciones_texto' con 3
#    textos de ejemplo (strings), incluyendo algunos hashtags (palabras que
#    empiezan por '#'). Inventa tú los textos.
#    Por ejemplo, con otros textos, así sería el RESULTADO final (esto NO es
#    la solución, inventa los tuyos):
#      ['Aprendiendo #Python y #NumPy', 'Nuevo #proyecto de #datos', 'Sin hashtags hoy']

publicaciones_texto = np.array([
    'Aprendiendo #Python y #NumPy',
    'Feliz cumpleanos a mi mejor amiga',
    'Nuevo proyecto de #datos con #IA y #python',
])

# 2. Define una función normal de Python 'contar_hashtags(texto)' que
#    reciba un string y devuelva cuántas palabras empiezan por '#'.
#    Pista:  separa el texto en una lista de palabras.
#    Por ejemplo, para el texto 'Nuevo #proyecto de #datos', la función
#    debe devolver 2.
# 3. Convierte 'contar_hashtags' en una función universal (ufunc) de NumPy:
#    una versión de la función que se puede aplicar directamente sobre un
#    array completo, elemento a elemento, sin necesidad de bucles.
#    Busca en la documentación de NumPy la función que transforma una
#    función normal de Python en una ufunc (necesitarás indicarle la
#    función original, cuántos argumentos de entrada tiene -1- y cuántos
#    valores de salida devuelve -1-). Guarda el resultado en
#    'ufunc_hashtags'.
# 4. Aplica 'ufunc_hashtags' sobre todo el array 'publicaciones_texto' de
#    una sola vez (sin bucles) y guarda el resultado en
#    'hashtags_por_publicacion'.
#    Por ejemplo, para un array de 3 textos con 2, 0 y 1 hashtags
#    respectivamente, el RESULTADO sería algo como [2, 0, 1].

def contar_hashtags(texto):
    return sum(1 for palabra in texto.split() if palabra.startswith('#'))


ufunc_hashtags = np.frompyfunc(contar_hashtags, 1, 1)
hashtags_por_publicacion = ufunc_hashtags(publicaciones_texto)
print(hashtags_por_publicacion)
````

````python
# ─── Ejercicio 8 — Normalización y estandarización ─────────────────────────
# 1. Normaliza el array 'edades_ejercicio' (ejercicio 1) entre 0 y 1 usando
#    la fórmula Min-Max:
#      valor_normalizado = (valor - minimo) / (maximo - minimo)
#    Guarda el resultado en 'edades_normalizadas'.
#    Por ejemplo, con otros datos, así sería el RESULTADO final (esto NO es
#    la solución): para el array [10, 20, 30], el resultado normalizado
#    sería [0.0, 0.5, 1.0].
#
# 2. Estandariza 'edades_ejercicio' calculando su Z-score:
#      z = (valor - media) / desviacion_estandar
#    Calcula la desviación estándar con un método de NumPy. Para evitar una
#    división por cero si la desviación estándar fuera 0, sustitúyela por 1
#    con un método de NumPy que permite elegir entre dos valores según se
#    cumpla o no una condición, antes de dividir. Guarda el resultado en
#    'edades_estandarizadas'.

minimo_ej8 = edades_ejercicio.min()
maximo_ej8 = edades_ejercicio.max()
edades_normalizadas = (edades_ejercicio - minimo_ej8) / (maximo_ej8 - minimo_ej8)

media_ej8 = edades_ejercicio.mean()
desviacion_ej8 = edades_ejercicio.std()
desviacion_segura_ej8 = np.where(desviacion_ej8 == 0, 1, desviacion_ej8)
edades_estandarizadas = (edades_ejercicio - media_ej8) / desviacion_segura_ej8
print(edades_normalizadas, edades_estandarizadas)
````

````python
# ─── Ejercicio A.1 — Puntuación de usuarios: normalizar y ordenar ─────────
# 1. A partir de 'usuarios_csv' (cargado arriba), crea tres arrays de NumPy:
#      'seguidores_a1'    -> el campo 'seguidores' de cada usuario, como float
#      'seguidos_a1'      -> el campo 'seguidos' de cada usuario, como float
#      'publicaciones_a1' -> el campo 'publicaciones_totales' de cada
#                            usuario, como float
#    Recuerda que estos campos llegan como texto desde el CSV: conviértelos
#    con float(). Los tres arrays deben tener un valor por usuario, en el
#    mismo orden que 'usuarios_csv'.
#
# 2. Calcula, SIN bucles, una puntuación por usuario con la fórmula:
#      puntuacion = 0.5 * seguidores + 0.2 * publicaciones - 0.1 * seguidos
#    Guárdala en 'puntuacion_a1'.
# 3. Normaliza 'puntuacion_a1' entre 0 y 1 con la fórmula Min-Max y
#    guárdala en 'puntuacion_normalizada_a1'.
# 4. Obtén los ÍNDICES de los usuarios ordenados de MAYOR a MENOR
#    puntuación, usando un método de NumPy que ordena y devuelve
#    posiciones en lugar de valores. Guárdalos en 'indices_orden_a1'.
#    Por ejemplo, con otros datos, así sería el RESULTADO para un array de
#    solo 4 puntuaciones [10, 40, 25, 5]: [1, 2, 0, 3] (esto NO es la
#    solución).

seguidores_a1 = np.array([float(u['seguidores']) for u in usuarios_csv])
seguidos_a1 = np.array([float(u['seguidos']) for u in usuarios_csv])
publicaciones_a1 = np.array([float(u['publicaciones_totales']) for u in usuarios_csv])

puntuacion_a1 = 0.5 * seguidores_a1 + 0.2 * publicaciones_a1 - 0.1 * seguidos_a1
minimo_a1 = puntuacion_a1.min()
maximo_a1 = puntuacion_a1.max()
puntuacion_normalizada_a1 = (puntuacion_a1 - minimo_a1) / (maximo_a1 - minimo_a1)
indices_orden_a1 = np.argsort(puntuacion_a1)[::-1]
print('Top 5 usuarios (índices):', indices_orden_a1[:5])
````

````python
# ─── Ejercicio A.2 — Matriz mes × tipo de contenido y totales por ejes ────
# 1. A partir de 'publicaciones_csv' (cargado arriba), construye una matriz
#    de NumPy llamada 'publicaciones_a2' con 12 filas (una por mes del año,
#    de enero a diciembre, independientemente del año) y una columna por
#    cada tipo de contenido distinto que exista en los datos, con el número
#    de publicaciones de cada tipo en cada mes. El campo 'fecha' tiene el
#    formato 'YYYY-MM-DD', así que puedes extraer el mes con fecha[5:7].
#    Guarda también, en el mismo orden que las columnas, la lista de tipos
#    de contenido en 'tipos_contenido_a2'.
#
# 2. Calcula el total de publicaciones de cada mes (suma por filas) y
#    guárdalo en 'totales_por_mes_a2'.
# 3. Calcula el total de publicaciones de cada tipo de contenido (suma por
#    columnas) y guárdalo en 'totales_por_tipo_a2'.
# 4. Obtén el ÍNDICE del mes con más publicaciones totales, usando un
#    método de NumPy que devuelve la posición del valor máximo. Guárdalo en
#    'mes_maximo_a2'.
# 5. Obtén el ÍNDICE del tipo de contenido más frecuente de la misma forma
#    y guárdalo en 'tipo_maximo_a2'.
#    Por ejemplo, con otra matriz más pequeña de 3 meses x 2 tipos, así
#    serían los RESULTADOS (esto NO es la solución): matriz
#    [[5, 2], [1, 9], [4, 4]] -> totales por mes [7, 10, 8], totales por
#    tipo [10, 15], mes_maximo 1, tipo_maximo 1.

from collections import Counter

tipos_contenido_a2 = sorted({p['tipo_contenido'] for p in publicaciones_csv})
conteo_a2 = Counter((int(p['fecha'][5:7]), p['tipo_contenido']) for p in publicaciones_csv)
publicaciones_a2 = np.array([
    [conteo_a2.get((mes, tipo), 0) for tipo in tipos_contenido_a2]
    for mes in range(1, 13)
])

totales_por_mes_a2 = publicaciones_a2.sum(axis=1)
totales_por_tipo_a2 = publicaciones_a2.sum(axis=0)
mes_maximo_a2 = int(np.argmax(totales_por_mes_a2))
tipo_maximo_a2 = int(np.argmax(totales_por_tipo_a2))
print('Tipos:', tipos_contenido_a2)
print('Mes con más publicaciones (0=enero):', mes_maximo_a2)
print('Tipo más frecuente:', tipos_contenido_a2[tipo_maximo_a2])
````

````python
# ─── Ejercicio A.3 — División entrenamiento/prueba y reunión ──────────────
# (Ejercicio de práctica libre: no forma parte de la entrega automática)
#
# 1. A partir de 'publicaciones_csv' (cargado arriba), construye una matriz
#    de NumPy llamada 'publicaciones_a3' con una fila por publicación y 2
#    columnas: 'likes' y 'comentarios' (conviértelos a float; recuerda que
#    llegan como texto con formato decimal).
# 2. Divide 'publicaciones_a3' en dos partes: 'entrenamiento_a3' con el 80%
#    de las filas (las primeras) y 'prueba_a3' con el 20% restante (las
#    últimas). Pista: puedes calcular cuántas filas tiene el 80% a partir
#    de publicaciones_a3.shape.
# 3. Vuelve a unir ambas partes con un método de NumPy y comprueba, con un
#    método de comparación de NumPy, que el resultado coincide con
#    'publicaciones_a3'.
# 4. Comprueba que el número de filas de 'entrenamiento_a3' más el de
#    'prueba_a3' coincide con el número de filas de 'publicaciones_a3'.

def a_float_seguro(valor):
    return float(valor) if valor else 0.0


publicaciones_a3 = np.array([
    [a_float_seguro(p['likes']), a_float_seguro(p['comentarios'])]
    for p in publicaciones_csv
])

n_filas_a3 = publicaciones_a3.shape[0]
n_entrenamiento_a3 = int(n_filas_a3 * 0.8)
entrenamiento_a3 = publicaciones_a3[:n_entrenamiento_a3]
prueba_a3 = publicaciones_a3[n_entrenamiento_a3:]

reunidas_a3 = np.vstack([entrenamiento_a3, prueba_a3])
coincide_con_original_a3 = np.array_equal(reunidas_a3, publicaciones_a3)
filas_coinciden_a3 = (entrenamiento_a3.shape[0] + prueba_a3.shape[0]) == n_filas_a3
print('Forma original:', publicaciones_a3.shape)
print('Entrenamiento/prueba:', entrenamiento_a3.shape, prueba_a3.shape)
print('¿Coincide al unir?', coincide_con_original_a3, '¿Filas coinciden?', filas_coinciden_a3)
````

````python
# ─── Ejercicio A.4 — Matriz de adyacencia y conexiones indirectas ─────────
# 1. A partir de 'usuarios_csv' y 'conexiones_json' (cargados arriba),
#    construye una matriz de adyacencia de NumPy llamada 'conexiones_a4',
#    de tamaño (num_usuarios x num_usuarios), donde la posición [i, j] valga
#    1 si el usuario de la fila i sigue al usuario de la fila j, y 0 en caso
#    contrario. Usa el mismo orden de usuarios que 'usuarios_csv' para filas
#    y columnas. Recuerda que en 'conexiones_json' los ids ya son enteros
#    (vienen de JSON, no de CSV).
#
# 2. Calcula cuántas conexiones DIRECTAS tiene cada usuario (a cuántos
#    usuarios sigue), sumando por filas. Guárdalo en
#    'conexiones_directas_a4'.
# 3. Calcula las conexiones INDIRECTAS de longitud dos ("amigos de amigos")
#    con un producto matricial entre 'conexiones_a4' y ella misma. El valor
#    en la posición [i, j] indica de cuántas formas el usuario i llega al
#    usuario j en exactamente dos pasos. Guarda el resultado en
#    'conexiones_indirectas_a4'.
#    Por ejemplo, con una red de solo 3 usuarios donde el 0 sigue al 1 y el
#    1 sigue al 2, así serían los RESULTADOS (esto NO es la solución):
#    conexiones_directas = [1, 1, 0], y la posición [0, 2] de las
#    conexiones indirectas valdría 1 (0 -> 1 -> 2).

ids_usuarios_a4 = [int(u['id_usuario']) for u in usuarios_csv]
indice_por_id_a4 = {id_usuario: indice for indice, id_usuario in enumerate(ids_usuarios_a4)}
n_usuarios_a4 = len(ids_usuarios_a4)

conexiones_a4 = np.zeros((n_usuarios_a4, n_usuarios_a4), dtype=int)
for registro in conexiones_json:
    origen = indice_por_id_a4.get(registro['id_usuario'])
    if origen is None:
        continue
    for id_destino in registro.get('siguiendo', []):
        destino = indice_por_id_a4.get(id_destino)
        if destino is not None:
            conexiones_a4[origen, destino] = 1

conexiones_directas_a4 = conexiones_a4.sum(axis=1)
conexiones_indirectas_a4 = conexiones_a4 @ conexiones_a4
print('Forma de la matriz de conexiones:', conexiones_a4.shape)
print('Conexiones directas (primeros 5 usuarios):', conexiones_directas_a4[:5])
````

````python
# ─── Ejercicio A.5 — Aleatoriedad reproducible ────────────────────────────
# 1. Crea un generador de números aleatorios reproducible a partir de la
#    semilla 42 (usa el método de NumPy pensado para generación aleatoria
#    moderna y reproducible a partir de una semilla).
# 2. Simula 30 publicaciones nuevas generando, EN ESTE ORDEN y con el mismo
#    generador, tres arrays de tamaño 30 con enteros aleatorios:
#      likes           -> entre 0 y 5000
#      comentarios     -> entre 0 y 500
#      visualizaciones -> entre 100 y 50000
# 3. Calcula el engagement de cada publicación, sin bucles, con la fórmula:
#      engagement = (likes + comentarios) / visualizaciones
#    Guárdalo en 'engagement_a5'.
# 4. Repite el proceso completo (pasos 1 a 3) con un NUEVO generador creado
#    con la MISMA semilla, y comprueba, con un método de NumPy que compara
#    arrays completos, si el resultado es idéntico al anterior. Guarda el
#    resultado booleano (True/False) en 'reproducible_a5'.

rng_a5 = np.random.default_rng(42)
likes_a5 = rng_a5.integers(0, 5000, size=30)
comentarios_a5 = rng_a5.integers(0, 500, size=30)
visualizaciones_a5 = rng_a5.integers(100, 50000, size=30)
engagement_a5 = (likes_a5 + comentarios_a5) / visualizaciones_a5

rng_a5_bis = np.random.default_rng(42)
likes_a5_bis = rng_a5_bis.integers(0, 5000, size=30)
comentarios_a5_bis = rng_a5_bis.integers(0, 500, size=30)
visualizaciones_a5_bis = rng_a5_bis.integers(100, 50000, size=30)
engagement_a5_bis = (likes_a5_bis + comentarios_a5_bis) / visualizaciones_a5_bis

reproducible_a5 = np.array_equal(engagement_a5, engagement_a5_bis)
print('¿Reproducible?', reproducible_a5)
````
